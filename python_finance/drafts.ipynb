```python
current_prices = {'BTC': 35000, 'ETH': 2400}

for sym in portfolio:
    pnl = 0
    for transaction in transactions:
        if transaction['symbol'] == sym:
            if transaction['type'] == 'buy':
                pnl -= transaction['quantity'] * transaction['price']
            elif transaction['type'] == 'sell':
                pnl += transaction['quantity'] * transaction['price']
    # Add current value of remaining holdings
    pnl += portfolio[sym]['quantity'] * current_prices[sym]
    portfolio[sym]['pnl'] = round(pnl, 2)

print(portfolio)
```

```python

# Часть 1

today_prices = {
    'BTC': 35000,
    'ETH': 2400
}

for transaction in transactions:
    symbol = transaction['symbol']
    if 'invested' not in portfolio[symbol]:
        portfolio[symbol]['invested'] = 0
    if transaction['type'] == 'buy':
        portfolio[symbol]['invested'] += transaction['quantity'] * transaction['price']
    elif transaction['type'] == 'sell':
        portfolio[symbol]['invested'] -= transaction['quantity'] * transaction['price']

print(portfolio)
```

```python

# Часть 2

for symbol in portfolio:
    current_value = portfolio[symbol]['quantity'] * today_prices[symbol]
    portfolio[symbol]['pnl'] = current_value - portfolio[symbol]['invested']

print(portfolio)
```
