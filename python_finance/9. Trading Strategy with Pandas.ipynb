```python
import pandas as pd
import datetime
```

```python
df = pd.read_csv('BTC-USDT.csv')
df
```

```python
df['date'] = pd.to_datetime(df['date'], unit='ms', utc=True)
df
```

```python
#setting the date column as the index of the dataframe
df.set_index('date', inplace=True)
df
```

```python
df['ma50'] = df['close'].rolling(50).mean()
df['ma200'] = df['close'].rolling(200).mean()
df
```

```python
df['buy_signal'] = False
df['sell_signal'] = False
df
```

```python
# Creating buy and sell signals based on moving averages
df.loc[df['ma50'] > df['ma200'], 'buy_signal'] = True
df.loc[df['ma50'] < df['ma200'], 'sell_signal'] = True
df
```

```python
df['sell_signal'].value_counts()
```

```python
balance = 1000
position = None

for index, row in df.iterrows():
    # Check for buy signal
    if position is None and row['buy_signal'] is True:
        buy_price = row['close']
        usd_size = balance
        position = {
            'buy_price': buy_price,
            'usd_size': usd_size
        }
        
        print(f'{index} - Bought Bitcoin for {usd_size} USD at {buy_price} USD')
        
    # Check for sell signal
    if position is not None and row['sell_signal'] is True:
        sell_price = row['close']
        trade_results = (sell_price - position['buy_price']) / position['buy_price'] 
        balance += trade_results * position['usd_size']
        print(f'{index} - Sold Bitcoin for {balance} USD at price of {sell_price} USD ({round(trade_results * 100, 2)}% profit)')
        position = None
        
        
    

```

```python
print('Our final balance is: ', round(balance, 2), 'USD')
```

```python
class BacktestCrossMA:
    def __init__(self) -> None:
        self.df = pd.DataFrame()
        
    def load_data(self, path):
        self.df = pd.read_csv(path)
        self.df['date'] = pd.to_datetime(self.df['date'], unit='ms')
        self.df = self.df.set_index(['date'])
        #del self.df['date']
        
    def populate_indicators(self):
        self.df['ma50'] = self.df['close'].rolling(50).mean()
        self.df['ma200'] = self.df['close'].rolling(200).mean()
    
    def populate_signals(self):
        self.df['buy_signal'] = False
        self.df['sell_signal'] = False
        self.df.loc[(self.df['ma50'] > self.df['ma200']), 'buy_signal'] = True
        self.df.loc[(self.df['ma50'] < self.df['ma200']), 'sell_signal'] = True
        
    def run_backtest(self):
        balance = 1000
        position = None
        asset = 'BTC'
        
        for index, row in self.df.iterrows():
            # Check for buy signal
            if position is None and row['buy_signal']:
                open_price = row['close']
                position = {
                    'open_price' : open_price,
                    'usd_size' : balance,
                }
                
                print(f'{index} - Buy for {balance}$ of {asset} at price : {open_price}$')
                
            elif position is not None and row['sell_signal']:
                close_price = row['close']
                trade_result = (close_price - position['open_price']) / position['open_price']
                balance = balance + trade_result * position['usd_size']
                position = None
                print(f'{index} - Sell for {balance}$ of {asset} at {close_price}$')
                
        print(f'Final balance: {balance}$')
        
```

```python
bt = BacktestCrossMA()
bt.load_data('BTC-USDT.csv')
bt.populate_indicators()
bt.populate_signals()
bt.run_backtest()
```
