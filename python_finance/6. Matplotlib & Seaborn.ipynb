```python
# Import pandas library
import pandas as pd

# Read the CSV file into a DataFrame
df = pd.read_csv('house_sales_in_king_county_usa.csv')
df
```

```python
df['price'].mean()
```

```python
# Create a histogram of the price data
plt.hist(df['price'], bins=100)
plt.xlabel('Price')
plt.ylabel('N houses')
plt.show()
```

```python
# Filtering out houses with price less than 2 million
plt.hist(df.loc[df['price'] < 1e6]['price'], bins=20)
plt.xlabel('Price')
plt.ylabel('N houses')
plt.show()
```

```python
# Here colors are used to represent the price of the house, and the size of the dots is set to 5. The 'inferno' colormap is used to represent the price values.
plt.scatter(df['sqft_living'], df['price'], s=5, c=df['price'], cmap='inferno')
plt.title('Price as a function of the living area')
plt.xlabel('Living area')
plt.ylabel('Price')
plt.show()
```

```python
# Showing by colors the number of bedrooms in the house, and filtering out houses with more than 6 bedrooms. The 'inferno' colormap is used to represent the number of bedrooms.

df6 = df.loc[df['bedrooms'] < 7] # Filtering out houses with more than 6 bedrooms


plt.figure(figsize=(8,6)) # Setting the figure size to 8x6 inches
scatter = plt.scatter(df6['sqft_living'], df6['price'], s=5, c=df6['bedrooms'], cmap='inferno') # 
plt.title('Price as a function of the living area')
plt.xlabel('Living area')
plt.ylabel('Price')

cbar = plt.colorbar(scatter) # Adding a colorbar to the scatter plot
cbar.set_label('N bedrooms') # Setting the label for the colorbar

plt.show()
```

```python
# Grouping the data by the 'view' column and calculating the mean price for each group. The result is stored in a new DataFrame called 'df_view'.

df_view = df.groupby('view')['price'].mean()
df_view
```

```python
# Creating a bar plot to visualize the mean price of houses based on the 'view' column.
# The x-axis represents the different views, and the y-axis represents the mean price for each view.

df_view = df.groupby('view')['price'].mean()
plt.bar(df_view.index, df_view.values, color='purple')
plt.show()
```

```python
# Grouping the data by month and calculating the mean price for each month. The result is stored in a new DataFrame called 'df_grouped'.
df['date'] = pd.to_datetime(df['date'])
df_grouped = df.resample('ME', on='date').mean()
df_grouped

# Creating a line plot to visualize the mean price of houses over time. The x-axis represents the date, and the y-axis represents the average price for each month.
plt.plot(df_grouped['price'])
plt.xlabel('Date')
plt.ylabel('Average Price')
plt.title('Mean Price of Houses Over Time')
plt.show()

```

```python
# Display the 5 Largest Sales Volumes by Year

df_year = df.groupby(df['date'].dt.year)['price'].sum()
df_year

plt.bar(df_year.index.astype(int), df_year.values, color='green')
plt.xticks(df_year.index.astype(int))
plt.xlabel('Year')
plt.ylabel('Total Sales Volume')
plt.title('Total Sales Volume by Year')
plt.show()
```

```python
df['date'] = pd.to_datetime(df['date'])

sales_by_year = df['date'].dt.year.value_counts().nlargest(5)
plt.figure(figsize=(10, 5))
sales_by_year.sort_index().plot(kind='bar')
plt.title('Top 5 Years with the Largest Sales Volumes')
plt.xlabel('Year')
plt.ylabel('Number of Sales')
plt.show()
```
