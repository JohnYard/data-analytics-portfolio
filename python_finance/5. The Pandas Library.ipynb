```python
# This will create a new column 'low_high_avg' in the DataFrame, which is the average of the 'low' and 'high' columns.
df['low_high_avg'] = (df['low'] + df['high'] / 2)
df
```

```python
# Calculating the mean, median, and standard deviation of the 'close' column in the DataFrame.
mean_prices = df["close"].mean()
median_prices = df["close"].median()
std_prices = df["close"].std()

# Printing the calculated statistics for the 'close' column.
print("Average closing prices:", round(mean_prices,2), '$')
print("Median closing prices:", median_prices)
print("Standard deviation of closing prices:", std_prices)
```

```python
# Creating a moving average of the closing prices with a window of 9 days
df['close_ma'] = df['close'].rolling(window=9).mean()
df
```

```python
# Checking the statistical summary of the DataFrame after adding the moving average column.
df.describe() 
```

```python
# Creating a new column 'daily_return' that calculates the daily return of the stock based on the 'close' price.

df['daily_return'] = df['close'].pct_change()  # This will calculate the daily return as the percentage change in the 'close' price from the previous day.
df['volatility'] = df['daily_return'].rolling(30).std()  # This will calculate the rolling standard deviation of the 'daily_return' over a window of 30 days, which is a measure of volatility.
df
```

```python
# Sorting the DataFrame by a column colatility in descending order to see the most volatile days at the top.

df.sort_values(by='daily_return', ascending=True, inplace=True) # This will sort the DataFrame by the 'daily_return' column in ascending order, modifying it in place.

df
```

```python
# Calculating the change between the 'close' and 'open' prices as a percentage of the 'close' price and storing it in a new column 'change'.
df['change'] = (df['close'] - df['open']) / df['close'] 
# This will sort the DataFrame by the 'change' column in ascending order, modifying it in place.
df.sort_values(by='change', ascending=True, inplace=True) 
df
```

```python
# Importing CVS into new df with pandas
df_t = pd.read_csv('TITANIC.csv')  # This will read the CSV file 'TITANIC.csv' into a new DataFrame df_t.

df_t
```

```python
# Looking at the first 10 rows of the Titanic DataFrame to get an overview of the data.
df_t.head(10)
```

```python
# Determine the dataset's dimensions (i.e. the number of rows and columns).

df_t.shape  # This will return a tuple representing the dimensions of the DataFrame, where the first element is the number of rows and the second element is the number of columns.
```

```python
# Count the number of passengers in each class (Pclass).

df_t['Pclass'].value_counts()  # This will return a Series containing counts of unique values in the 'Pclass' column, effectively counting the number of passengers in each class.
```

```python
# This will calculate the average age of passengers in the Titanic dataset.
average_age = df_t['Age'].mean()  
average_age 
```

```python
# Sorting the Titanic DataFrame by the 'Age' column in descending order to find the oldest passengers.
df_t.sort_values(by='Age', ascending=False, inplace=True)
df_t

#Oldest - Barkworth, Mr. Algernon Henry Wilson
```

```python
# Sorting the Titanic DataFrame by the 'Age' column in ascending order to find the youngest passengers.
df_t.sort_values(by='Age', ascending=True, inplace=True)
df_t

# Youngest - Thomas, Master. Assad Alexander
```

```python
# Counting the number of survivors and non-survivors in the Titanic dataset.
df_t['Survived'].value_counts()
```

```python
# Counting the number of male and female passengers in the Titanic dataset.
df_t['Sex'].value_counts()
```

```python
# Calculate How many men survived
men_survived = df_t[(df_t['Sex'] == 'male') & (df_t['Survived'] == 1)].shape[0]  # This will count the number of male passengers who survived.
men_survived
```

```python
# Calculate How many women survived
women_survived = df_t[(df_t['Sex'] == 'female') & (df_t['Survived'] == 1)].shape[0]
women_survived
```

```python
# Other way to calculate how many women survived using len() function instead of shape[0].
women_survived1 = len(df_t[(df_t['Sex'] == 'female') & (df_t['Survived'] == 1)])
women_survived1
```

```python
# Calculate the survival rate for each passenger class (Pclass) in the Titanic dataset.
first_class_survival_rate = df_t[df_t['Pclass'] == 1]['Survived'].mean()  
print(first_class_survival_rate)

second_class_survival_rate = df_t[df_t['Pclass'] == 2]['Survived'].mean()
print(second_class_survival_rate)

third_class_survival_rate = df_t[df_t['Pclass'] == 3]['Survived'].mean()
print(third_class_survival_rate)
```

```python
# Calculating the total number of passengers in first class and the number of survivors in first class to compute the survival rate.
first_class_total = len(df_t['Pclass'] == 1)
print(first_class_total)

first_class_survived = len(df_t[(df_t['Pclass'] == 1) & (df_t['Survived'] == 1)])
print(first_class_survived)

first_class_survival_rate2 = first_class_survived / first_class_total
print(first_class_survival_rate2)

```

```python
# Selecting the 'Survived' column for passengers in first class to analyze their survival status.
first_class_survival_rate = df_t[df_t['Pclass'] == 1]['Survived']
first_class_survival_rate
```

```python
# Creating a new column 'FamilySize' in the Titanic DataFrame that represents the total number of family members aboard (siblings/spouses + parents/children).
df_t['FamilySize'] = df_t['SibSp'] + df_t['Parch']
df_t
```

```python
# Sorting the Titanic DataFrame by the 'FamilySize' column in descending order to find passengers
# who did not survive and had the largest families aboard.
df_t[df_t['Survived'] == 0].sort_values(by='FamilySize', ascending=False)


```
