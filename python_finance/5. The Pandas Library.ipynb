```python
mean_prices = df["close"].mean()
median_prices = df["close"].median()
std_prices = df["close"].std()

print("Average closing prices:", round(mean_prices,2), '$')
print("Median closing prices:", median_prices)
print("Standard deviation of closing prices:", std_prices)
```

```python
# Creating a moving average of the closing prices with a window of 9 days

df['close_ma'] = df['close'].rolling(window=9).mean()
df
```

```python
df.describe()  # 
```

```python
# Creating a new column 'daily_return' that calculates the daily return of the stock based on the 'close' price.

df['daily_return'] = df['close'].pct_change()  # This will calculate the daily return as the percentage change in the 'close' price from the previous day.
df['volatility'] = df['daily_return'].rolling(30).std()  # This will calculate the rolling standard deviation of the 'daily_return' over a window of 30 days, which is a measure of volatility.
df
```

```python
# Sorting the DataFrame by a column colatility in descending order to see the most volatile days at the top.

df.sort_values(by='daily_return', ascending=True, inplace=True) # This will sort the DataFrame by the 'daily_return' column in ascending order, modifying it in place.

df
```

```python
df['change'] = (df['close'] - df['open']) / df['close'] 
df.sort_values(by='change', ascending=True, inplace=True) # This will sort the DataFrame by the 'change' column in ascending order, modifying it in place.

df
```
