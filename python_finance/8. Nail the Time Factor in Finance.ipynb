```python
# Import library for date and time manipulation
import datetime 
```

```python
# Showing difference between datetime.date and string representation of date
print(datetime.date(2024,1,2))
print('2024-01-02')
```

```python
# Creating a string representation of a date and converting it to a datetime object
d1 = '2024-01-02'

# Converts string to datetime object
print(datetime.datetime.strptime(d1, '%Y-%m-%d')) 
```

```python
# String with slashes instead of dashes and time
d2 = '2024/01/02 09:35:02' 

# Converting string to datetime object with slashes instead of dashes and current time
print(datetime.datetime.strptime(d2, '%Y/%m/%d %H:%M:%S')) 
```

```python
# Importing the time module to save the current time in seconds since the epoch (January 1, 1970)
import time

# Prints the current time in seconds since the epoch (January 1, 1970)
print(time.time()) 
```

```python
# Saving the current time in seconds since the epoch (January 1, 1970) to a variable
t1 = time.time()

# Prints the current time in seconds since the epoch (January 1, 1970)
print(t1) 

# Converts seconds since epoch to datetime object
print(datetime.datetime.fromtimestamp(t1)) 
```

```python
# Creating a string representation of a date and time and converting it to a datetime object
d2 = '2024/01/02 09:35:02' 

# Converting string to datetime object with and time
d2_datetime = datetime.datetime.strptime(d2, '%Y/%m/%d %H:%M:%S') 

# Prints the datetime object
print(d2_datetime) 

# Converts datetime object to seconds since epoch
print(d2_datetime.timestamp()) 
```

```python
# Importing the pandas library for data manipulation and analysis
import pandas as pd
```

```python
# Reading a CSV file containing Bitcoin to USDT data into a pandas DataFrame
df = pd.read_csv('BTC-USDT.csv')
df
```

```python
# Converts the 'date' column from milliseconds since epoch to datetime objects in UTC
df['date'] = pd.to_datetime(df['date'], unit='ms', utc=True)
df

# Prints the data type of the 'date' column
print(df['date'].dtype) 
```

```python
# Creating a new column 'weekday' in the DataFrame that contains the day of the week for each date (0=Monday, 6=Sunday)
df['weekday'] = df['date'].dt.weekday
df
```

```python
# Creating a mapping of numeric weekday values to their corresponding day names
day_name_map = {
    0: 'Monday',
    1: 'Tuesday',
    2: 'Wednesday',
    3: 'Thursday',
    4: 'Friday',
    5: 'Saturday',
    6: 'Sunday'
}

# Maps the numeric weekday values to their corresponding day names using the day_name_map dictionary
df['weekday'] = df['weekday'].map(day_name_map) 
df
```

```python
# Calculates the difference between the current date and the previous date in the 'date' column
df['date_diff'] = df['date'] - df['date'].shift(1) 
df
```

```python
# Counts the occurrences of each unique value in the 'date_diff' column
df['date_diff'].value_counts()
```

```python
# Filters the DataFrame to show rows where the 'date_diff' column is equal to '0 days 00:00:00'
print(df.loc[df['date_diff'] == '0 days 00:00:00']) 
```

```python
# Groups the DataFrame by the 'weekday' column and calculates the mean of the 'volume' column for each group
df.groupby('weekday').agg({'volume':'mean'}) 
```

```python
# Creating a new column nyse (New York Stock Exchange) that consists of True or False values based on whether the time 
# in the 'date' column falls within the NYSE trading hours (9:30 AM to 4:00 PM Eastern Time)
df['nyse'] = df['date'].apply(lambda x: x.time() >= datetime.time(9, 30) and x.time() <= datetime.time(16, 0))
df
```

```python
# Vectorized dt.time version of the above code, which is faster and more efficient than using apply() with a lambda function.
# Creating a new column 'date_nyse' that converts the 'date' column to New York time zone
df['date_nyse'] = df['date'].dt.tz_convert('America/New_York')

# Extracting the time component from the 'date_nyse' column and storing it in a new variable 'time_col'
time_col = df['date_nyse'].dt.time

# Creating a column 'nyse' that consists of True or False values based on whether the time in the 'date' column falls within the NYSE trading hours
# (9:30 AM to 4:00 PM Eastern Time)
df['nyse'] = (time_col >= datetime.time(9, 30)) & (time_col <= datetime.time(16, 0))
df

# Creating a column 'euronext' that consists of True or False values based on whether the time 
# in the 'date' column falls within the Euronext trading hours 
# (9:00 AM to 5:30 PM Central European Time) but in American Eastern Time (CET is 6 hours ahead of EST, 
# so the trading hours in EST are 3:00 AM to 11:30 AM)
df['euronext'] = (time_col >= datetime.time(3, 0)) & (time_col <= datetime.time(11, 30))
df

# Creating a column 'tokyo' that consists of True or False values based on whether the time 
# in the 'date' column falls within the Tokyo Stock Exchange trading hours
# (9:00 AM to 3:00 PM Japan Standard Time) but in American Eastern Time (JST is 13 hours ahead of EST, 
# so the trading hours in EST are 8:00 PM to 2:00 AM)
# Since this range wraps around midnight, we need to check if the time is >= 20:00 OR <= 02:00, instead of using between() which assumes start <= end
df['tokyo'] = (time_col >= datetime.time(20, 0)) | (time_col <= datetime.time(2, 0))
df


```

```python
# Compare columns 'nyse', 'euronext', and 'tokyo' by mean volume during days of the week when each market is open. 

df.groupby('weekday').agg({
    'nyse': 'mean',
    'euronext': 'mean',
    'tokyo': 'mean'
})
```

```python
# Alternatively, you can use the following code to compare the mean volume for each market during days of the week when each market is open:
print(df[df['nyse'] == True].groupby('weekday').agg({'volume': 'mean'}))

print(df[df['euronext'] == True].groupby('weekday').agg({'volume': 'mean'}))

print(df[df['tokyo'] == True].groupby('weekday').agg({'volume': 'mean'}))

```

```python
# Alternative option to compare the mean volume for each market during days of the week when each market is open:
result = pd.DataFrame({
    'nyse': df[df['nyse']]['volume'].groupby(df['weekday']).mean(),
    'euronext': df[df['euronext']]['volume'].groupby(df['weekday']).mean(),
    'tokyo': df[df['tokyo']]['volume'].groupby(df['weekday']).mean()
})

print(result)
```
